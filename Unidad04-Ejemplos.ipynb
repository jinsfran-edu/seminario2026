## Cifrado simetrico

```python
from cryptography.fernet import Fernet

def generate_key():
    """
    Generates a secure random key for symmetric encryption.
    Returns:
        bytes: The generated key.
    """
    return Fernet.generate_key()

def encrypt_message(key, message):
    """
    Encrypts a message using the provided symmetric key.
    
    Args:
        key (bytes): The symmetric encryption key.
        message (str): The plaintext message to encrypt.
    
    Returns:
        bytes: The encrypted ciphertext.
    """
    if not isinstance(message, str):
        raise TypeError("Message must be a string.")
    fernet = Fernet(key)
    return fernet.encrypt(message.encode())

def decrypt_message(key, token):
    """
    Decrypts a ciphertext using the provided symmetric key.
    
    Args:
        key (bytes): The symmetric encryption key.
        token (bytes): The encrypted ciphertext.
    
    Returns:
        str: The decrypted plaintext message.
    """
    fernet = Fernet(key)
    return fernet.decrypt(token).decode()

if __name__ == "__main__":
    try:
        # Step 1: Generate a key
        key = generate_key()
        print(f"Generated Key: {key.decode()}")

        # Step 2: Encrypt a message
        plaintext = "Mensaje secreto"
        ciphertext = encrypt_message(key, plaintext)
        print(f"Encrypted: {ciphertext}")

        # Step 3: Decrypt the message
        decrypted = decrypt_message(key, ciphertext)
        print(f"Decrypted: {decrypted}")

    except Exception as e:
        print(f"Error: {e}")
```

## Cifrado asimétrico
### RSA

```python
from cryptography.hazmat.primitives.asymmetric import rsa, padding
from cryptography.hazmat.primitives import serialization, hashes

# Generate RSA key pair
private_key = rsa.generate_private_key(public_exponent=65537, key_size=2048)
public_key = private_key.public_key()

print(
    f"Private Key:\n {private_key.private_bytes(
        encoding=serialization.Encoding.PEM,
        format=serialization.PrivateFormat.TraditionalOpenSSL,
        encryption_algorithm=serialization.NoEncryption()
    ).decode()} \n"
)
print (f"Public Key:\n {public_key.public_bytes(encoding=serialization.Encoding.PEM, format=serialization.PublicFormat.SubjectPublicKeyInfo).decode()}\n")

# Encrypt
message = b"Mensaje secreto"
ciphertext = public_key.encrypt(
    message,
    padding.OAEP(
        mgf=padding.MGF1(algorithm=hashes.SHA256()),
        algorithm=hashes.SHA256(),
        label=None
    )
)
print (f"Mensaje cifrado: {ciphertext.hex()}")
# Decrypt
plaintext = private_key.decrypt(
    ciphertext,
    padding.OAEP(
        mgf=padding.MGF1(algorithm=hashes.SHA256()),
        algorithm=hashes.SHA256(),
        label=None
    )
)
print (f"Mensaje descifrado: {plaintext.decode()}")
```

## Hash

```python
import bcrypt
import hashlib

password = 'Prueba1234'

bytes = password.encode('utf-8')
salt = bcrypt.gensalt()



md5_hash = hashlib.md5(bytes).hexdigest()
sha1_hash = hashlib.sha1(bytes).hexdigest()
sha256_hash = hashlib.sha256(bytes).hexdigest()
sha512_hash = hashlib.sha512(bytes).hexdigest()
sha3_hash = hashlib.sha3_256(bytes).hexdigest()
sha3_512_hash = hashlib.sha3_512(bytes).hexdigest()

pbkdf2_hash = hashlib.pbkdf2_hmac('sha256', bytes, salt, 100000).hex()

hash = bcrypt.hashpw(bytes, salt)

print("MD5 Hash     : " + md5_hash)
print("SHA1 Hash    : " + sha1_hash)
print("SHA256 Hash  : " + sha256_hash)
print("SHA512 Hash  : " + sha512_hash)
print("SHA3 Hash    : " + sha3_hash)
print("SHA3_512 Hash: " + sha3_512_hash)
print("PBKDF2 Hash  : " + pbkdf2_hash)
print("BCrypt Hash  : " + str(hash, 'utf-8'))
```
