Con las funciones vistas hasta ahora, podemos obtener la cantidad de letras e en los apellidos de los empleados

````python
%%sql
SELECT IDEmpleado, Apellido,
LEN(Apellido) - LEN(REPLACE(Apellido, 'e', '')) AS numoccur
FROM Empleados
````

Reemplazar “.x.” por “.y.”. Mi cadena de inicial es “.x.x.x.x.”

````python
%%sql
SELECT REPLACE('.x.x.x.x.', '.x.', '.y.');
````

````python
%%sql
SELECT REPLACE(REPLACE('.x.x.x.x.', '.x.', '.y.'), '.x.', '.y.');
````

Reemplazar múltiples espacios contiguos con un único espacio:
```
'esta   es     una   cadena    con     muchos    espacios'

````python
%%sql
SELECT REPLACE(REPLACE(REPLACE('esta   es     una   cadena    con     muchos    espacios', ' ', '~ '), ' ~', ''), '~ ', ' ');
````

````python
%%sql
SELECT REPLACE('esta   es     una   cadena    con     muchos    espacios', ' ', '~ ');
````

````python
%%sql
SELECT REPLACE(REPLACE('esta   es     una   cadena    con     muchos    espacios', ' ', '~ '), ' ~', '');
````

Eliminar ceros a izquierda en una cadena, por ejemplo “0000001709”

````python
%%sql
SELECT TRIM(LEADING '0' FROM '0000001709');
````

````python
%%sql
SELECT REPLACE(LTRIM(REPLACE('0000001709', '0', ' ')), ' ', '0');
````

¿Qué pasa si la cadena original tiene espacios “0000001709 ABC”?

````python
%%sql
SELECT REPLACE(REPLACE(LTRIM(REPLACE(REPLACE('0000001709 ABC', ' ', '~'), '0', ' ')), ' ', '0'), '~', ' ');
````

````python
%%sql
SELECT SUBSTRING('0000001709', PATINDEX('%[^0]%', '0000001709'))
````

Verificar que una cadena contiene sólo dígitos numéricos

````python
%%sql
SELECT
  CASE
    WHEN '1759' LIKE REPLICATE('[0-9]', LEN('1759'))
      THEN 'Verdadero'
    ELSE 'Falso o Desconocido'
  END;
````

````python
%%sql
SELECT
CASE
WHEN '1759' NOT LIKE '%[^0-9]%' THEN 'Verdadero'
ELSE 'Falso o Desconocido'
END;
````

````python
%%sql
SELECT 
    CASE
    WHEN ISNUMERIC('1759') = 1
      THEN 'Verdadero'
    ELSE 'Falso o Desconocido'
  END;
````

Con la siguiente consulta, podemos ver los archivos que está utilizando la base de datos

````python
%%sql
SELECT physical_name 
FROM [sys].[database_files]
````

Quiero obtener los nombres de los archivos, sin la ruta completa

````python
%%sql
SELECT RIGHT(physical_name, CHARINDEX('/', REVERSE(physical_name))-1)
FROM sys.database_files
````

Y sin usar RIGHT?

````python
%%sql
SELECT REVERSE(SUBSTRING(REVERSE(physical_name), 1, CHARINDEX('/', REVERSE(physical_name))-1))
FROM sys.database_files
````

Y para que funcione en Windows o Linux?

````python
%%sql
SELECT REVERSE(SUBSTRING(REVERSE(physical_name), 1, CHARINDEX('/', REVERSE(REPLACE(physical_name, '\', '/')))-1))
FROM sys.database_files
````
