```python
%%sql
CREATE TABLE #Usuarios (
    username NVARCHAR(50) PRIMARY KEY,
    password NVARCHAR(255) NOT NULL
);

INSERT INTO #Usuarios (username, password) VALUES ('admin', 'admin123');
INSERT INTO #Usuarios (username, password) VALUES ('user1', 'password1');

SELECT * FROM #Usuarios WHERE username = 'admin' AND password = 'admin123';
```

## Consulta ejemplo

```python
%%sql
SELECT IDEmpleado, YEAR(FechaPedido) AS aniopedido, COUNT(*) AS numpedidos
FROM Pedidos
WHERE IDCliente = 'SAVEA'
GROUP BY IDEmpleado, YEAR(FechaPedido)
HAVING COUNT(*) > 1
ORDER BY IDEmpleado, aniopedido;
```
