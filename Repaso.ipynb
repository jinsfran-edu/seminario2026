```python
%load_ext sql

import os
from dotenv import load_dotenv

load_dotenv()

%sql mssql://{os.getenv('User')}:{os.getenv('Password')}@{os.getenv('Server')}/{os.getenv('Database')}?encrypt=yes&trustServerCertificate=yes&connectionTimeout=30&driver=ODBC+Driver+18+for+SQL+Server

%sql SELECT @@VERSION AS 'SQL Server Version';
```

```python
%%sql
SELECT IDEmpleado, YEAR(FechaPedido) AS aniopedido, COUNT(*) AS numpedidos
FROM Pedidos
WHERE IDCliente = 'SAVEA'
GROUP BY IDEmpleado, YEAR(FechaPedido)
HAVING COUNT(*) > 1
ORDER BY IDEmpleado, aniopedido;
```
