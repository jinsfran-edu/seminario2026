## Usando variables

```sql
USE Pampero;
DECLARE @maxid AS INT = (SELECT MAX(IDPedido)
FROM Pedidos);
SELECT IDPedido, FechaPedido, IDEmpleado, IDCliente
FROM Pedidos
WHERE IDPedido = @maxid;
```

## Usando Subconsulta

```sql
SELECT IDPedido, FechaPedido, IDEmpleado, IDCliente
FROM Pedidos
WHERE IDPedido = (SELECT MAX(O.IDPedido)
FROM Pedidos AS O);
```

## Pedidos de empleados apellido empieza con C

```sql
SELECT IDPedido
FROM Pedidos
WHERE IDEmpleado =
(SELECT E.IDEmpleado
FROM Empleados AS E
WHERE E.Apellido LIKE 'C%');
```

## Pedidos de empleados apellido empieza con D

```sql
SELECT IDPedido
FROM Pedidos
WHERE IDEmpleado =
(SELECT E.IDEmpleado
FROM Empleados AS E
WHERE E.Apellido LIKE 'D%');
```

## Subconsulta Correlacionada
Obtener para cada cliente, IDCliente, y los siguientes datos del último pedido generado (IDPedido, FechaPedido, IDEmpleado). (el último pedido es el que tenga mayor IDPedido)

```sql
SELECT IDCliente, IDPedido, FechaPedido, IDEmpleado
FROM Pedidos AS P1
WHERE IDPedido =
(SELECT MAX(P2.IDPedido)
FROM Pedidos AS P2
WHERE P2.IDCliente = P1.IDCliente);
```

## EXISTS

Obtener los clientes de España que han realizado pedidos:

```sql
SELECT IDCliente, NombreEmpresa
FROM Clientes AS C
WHERE Pais = 'España'
AND EXISTS
(SELECT * FROM Pedidos AS P
WHERE P.IDCliente = C.IDCliente);
```

## NOT EXISTS vs. NOT IN
Necesitamos los clientes de España que no realizaron pedidos:

```sql
SELECT IDCliente, NombreEmpresa
FROM Clientes AS C
WHERE pais = 'España'
AND NOT EXISTS
(SELECT * FROM Pedidos AS P
WHERE P.IDCliente = C.IDCliente);
```

```sql
SELECT IDCliente, NombreEmpresa
FROM Clientes AS C
WHERE pais = 'España'
AND IDCliente NOT IN
(SELECT IDCliente FROM Pedidos AS P);
```

## Division Relacional

Supongamos que necesito el código de los clientes que fueron atendidos por todos los empleados de Estados Unidos (EE.UU.)

Abusando del conocimiento de los datos podría utilizar lo siguiente:

```sql
SELECT IDCliente 
FROM Pedidos
WHERE IDEmpleado IN (1, 2, 3, 4, 8)
GROUP BY IDCliente
HAVING COUNT(DISTINCT IDEmpleado) = 5;
```

```sql
SELECT IDCliente 
FROM Pedidos
WHERE IDEmpleado IN 
    (SELECT IDEmpleado 
    FROM Empleados 
    WHERE Pais = 'EE.UU.')
    GROUP BY IDCliente
    HAVING COUNT(DISTINCT IDEmpleado) = 
                                        (SELECT COUNT(*) 
                                        FROM Empleados
                                        WHERE Pais = 'EE.UU.');
```

```sql
SELECT IDCliente FROM Clientes AS C
WHERE NOT EXISTS
    (SELECT * FROM Empleados AS E
    WHERE pais = 'EE.UU.'
    AND NOT EXISTS
        (SELECT * FROM Pedidos AS P
        WHERE P.IDCliente = C.IDCliente
        AND P.IDEmpleado = E.IDEmpleado));
```

## Expresiones de Tabla
### Tablas Derivadas


```sql
SELECT *
FROM (SELECT IDCliente, NombreEmpresa
        FROM Clientes
        WHERE Pais = 'EE.UU.') AS USAClies;
```

```sql
SELECT YEAR(FechaPedido) AS ordenanio, COUNT(DISTINCT IDCliente) AS numclies
FROM Pedidos
GROUP BY ordenanio;
```

```sql
SELECT ordenanio, 
COUNT(DISTINCT IDCliente) AS numclies
FROM (SELECT YEAR(FechaPedido) AS ordenanio, IDCliente
FROM Pedidos) AS D
GROUP BY ordenanio;
```

```sql
SELECT ordenanio, COUNT(DISTINCT IDCliente) AS numclies
FROM (SELECT YEAR(FechaPedido), IDCliente
        FROM Pedidos) AS D(ordenanio, IDCliente)
GROUP BY ordenanio;
```

```sql
SELECT ordenanio, numcusts
FROM (SELECT ordenanio, COUNT(DISTINCT IDCliente) AS numcusts
        FROM (SELECT YEAR(FechaPedido) AS ordenanio, IDCliente
                FROM Pedidos) AS D1
        GROUP BY ordenanio) AS D2
WHERE numcusts > 70;
```

```sql
SELECT Cur.ordenanio AS Año, Cur.numcusts AS ClientesAñoActual, Prv.numclies AS ClientesAñoAnterior, Cur.numcusts - Prv.numclies AS Crecimiento
FROM (SELECT YEAR(FechaPedido) AS ordenanio, COUNT(DISTINCT IDCliente) AS numcusts
FROM Pedidos GROUP BY YEAR(FechaPedido)) AS Cur
LEFT OUTER JOIN
(SELECT YEAR(FechaPedido) AS ordenanio, COUNT(DISTINCT IDCliente) AS numclies
FROM Pedidos GROUP BY YEAR(FechaPedido)) AS Prv
ON Cur.ordenanio = Prv.ordenanio + 1;
```

### CTE

```sql
WITH UsaClies AS
(
SELECT IDCliente, NombreEmpresa
FROM Clientes
WHERE Pais = 'EE.UU.'
)
SELECT * FROM UsaClies;
```

```sql
WITH C AS
(
SELECT YEAR(FechaPedido) AS ordenanio, IDCliente
FROM Pedidos
)
SELECT ordenanio, COUNT(DISTINCT IDCliente) AS numcusts
FROM C
GROUP BY ordenanio;
```

```sql
WITH C(ordenanio, IDCliente) AS
(
SELECT YEAR(FechaPedido), IDCliente
FROM Pedidos
)
SELECT ordenanio, COUNT(DISTINCT IDCliente) AS numcusts
FROM C
GROUP BY ordenanio;
```

```sql
WITH C1 AS
(
SELECT YEAR(FechaPedido) AS ordenanio, IDCliente
FROM Pedidos
),
C2 AS
(
SELECT ordenanio, COUNT(DISTINCT IDCliente) AS numclies
FROM C1
GROUP BY ordenanio
)
SELECT ordenanio, numclies
FROM C2
WHERE numclies > 70;
```

```sql
WITH CuentaAnual AS
(
SELECT YEAR(FechaPedido) AS ordenanio,
COUNT(DISTINCT IDCliente) AS numclies
FROM Pedidos
GROUP BY YEAR(FechaPedido)
)
SELECT Cur.ordenanio,
Cur.numclies AS curnumcusts, Prv.numclies AS prvnumcusts,
Cur.numclies - Prv.numclies AS growth
FROM CuentaAnual AS Cur
LEFT OUTER JOIN CuentaAnual AS Prv
ON Cur.ordenanio = Prv.ordenanio + 1;
```

```sql
--Consulta Recursiva para listar al empleado y todos los subordinados en el organigrama hasta el mas bajo en la piramide
WITH EmpCTE (IDEmpleado, Nombre, Apellido, puesto, jefeid)
AS
(
SELECT IDEmpleado, Nombre, Apellido, Puesto, JefeID
FROM Empleados
WHERE IDEmpleado = 5
UNION ALL
--RECURSIVIDAD
SELECT e.IDEmpleado, e.Nombre, e.Apellido, e.Puesto, e.JefeID
FROM Empleados AS e JOIN EmpCTE AS m
ON e.JefeID = m.IDEmpleado
)
-- Desplegar ResultSet
SELECT * FROM EmpCTE;
```

```sql
--Consulta Recursiva para listar al empleado y todos los subordinados en el organigrama hasta el mas bajo en la piramide
WITH EmpCTE (IDEmpleado, Nombre, Apellido, puesto, jefeid)
AS
(
SELECT IDEmpleado, Nombre, Apellido, Puesto, JefeID
FROM Empleados
WHERE IDEmpleado = 5
UNION ALL
--RECURSIVIDAD
SELECT e.IDEmpleado, e.Nombre, e.Apellido, e.Puesto, e.JefeID
FROM Empleados AS e JOIN EmpCTE AS m
ON e.IDEmpleado = m.jefeid
)
-- Desplegar ResultSet
SELECT * FROM EmpCTE;
```

### Vistas

```sql
DROP VIEW IF EXISTS ClientesUSA;
GO
CREATE VIEW ClientesUSA
AS
SELECT
IDCliente, NombreEmpresa, NombreContacto, PuestoContacto, Direccion,
Ciudad, Region, CodigoPostal, Pais, Telefono, Fax
FROM Clientes
WHERE Pais = 'EE.UU.';
GO
```

### Funciones de Tabla en Línea

```sql
DROP FUNCTION IF EXISTS GetClieOrdenes;
GO
CREATE FUNCTION GetClieOrdenes
(@cid AS char(5)) RETURNS TABLE
AS
RETURN
SELECT IDPedido, IDCliente, IDEmpleado, FechaPedido, FechaRequerida,
FechaEnvio, EnvioPor, Flete, NombreEnvio, DireccionEnvio, CiudadEnvio,
RegionEnvio, CodigoPostalEnvio, PaisEnvio
FROM Pedidos
WHERE IDCliente = @cid;
GO
```

```sql
SELECT IDPedido, IDCliente
FROM GetClieOrdenes('ALFKI') AS O;
```

```sql
SELECT O.IDPedido, O.IDCliente, DP.IDProducto, DP.Cantidad
FROM GetClieOrdenes('ALFKI') AS O
JOIN [Detalles Pedido] AS DP
ON O.IDPedido = DP.IDPedido;
```
