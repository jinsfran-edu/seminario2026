### WITH TIES

```python
%%sql 
DROP TABLE IF EXISTS #podio;

CREATE TABLE #podio ( 
nombre nvarchar(50) PRIMARY KEY, 
pais nvarchar(50), 
tiempo numeric(18,2) 
); 

INSERT INTO #podio (nombre, pais, tiempo)
VALUES 
('Aleksandr Sadovnikov', 'RUS', 51.84),
('Chad Le Clos', 'RSA', 51.14),
('Joseph Schooling', 'SGP', 50.39),
('Laszlo Cseh', 'HUN', 51.14),
('Medhy Metella', 'FRA', 51.58),
('Michael Phelps', 'USA', 51.14),
('Tom Shields', 'USA', 51.73),
('Zhuhao Li', 'CHN', 51.26);

SELECT TOP (3) WITH TIES nombre, pais, tiempo FROM #podio ORDER BY tiempo ASC;
```

## Precedencia

```python
%%sql
SELECT IDPedido, IDCliente, 
IDEmpleado, FechaPedido
FROM Pedidos
WHERE IDCliente = 'ALFKI' 
AND IDEmpleado IN(1, 3, 5) 
OR IDCliente = 'VINET' 
AND IDEmpleado IN(2, 4, 6);
```

## CASE

```python
%%sql 
SELECT TOP(10) IDProducto, NombreProducto, IDCategoria,
CASE IDCategoria
WHEN 1 THEN 'Bebidas'
WHEN 2 THEN 'Condimentos'
WHEN 3 THEN 'Confites'
WHEN 4 THEN 'Lacteos'
WHEN 5 THEN 'Granos/Cereales'
WHEN 6 THEN 'Avicola'
WHEN 7 THEN 'Frutas'
WHEN 8 THEN 'Mariscos'
ELSE 'Categoria Desconocida'
END AS NombreCategoria
FROM Productos
```

```python
%%sql
SELECT TOP(10) IDPedido, IDCliente, val,
CASE
WHEN val < 1000.00 THEN 'Menos de 1000'
WHEN val BETWEEN 1000.00 AND 3000.00 THEN 'Entre 1000 y 3000'
WHEN val > 3000.00 THEN 'Mas de 3000'
ELSE 'Desconocido'
END AS CategoriaValor
FROM ValoresPedido
```

## GREATEST y LEAST

```python
%%sql
SELECT IDPedido, FechaRequerida, FechaEnvio,
GREATEST(FechaRequerida, FechaEnvio) AS fechatardia,
LEAST(Fecharequerida, FechaEnvio) AS fechatemprana
FROM Pedidos
WHERE IDCliente = 'BOLID';
```

```python
%%sql
SELECT IDCliente, NombreEmpresa
FROM Clientes
ORDER BY IDCliente
OFFSET 13 ROWS FETCH NEXT 1 ROWS ONLY;
```

```python
%%sql
SELECT IDCliente, NombreEmpresa
FROM Clientes
ORDER BY IDCliente COLLATE Traditional_Spanish_100_CI_AI_SC_UTF8
OFFSET 13 ROWS FETCH NEXT 1 ROWS ONLY;
```

```python
%%sql
SELECT IDCliente, NombreEmpresa, OrdMod, OrdTrad
FROM (SELECT IDCliente, NombreEmpresa,
ROW_NUMBER() OVER 
(ORDER BY IDCliente COLLATE Traditional_Spanish_100_CI_AI_SC_UTF8) AS OrdTrad,
ROW_NUMBER() OVER 
(ORDER BY IDCliente COLLATE Modern_Spanish_100_CI_AI_SC_UTF8) AS OrdMod
FROM Clientes) AS D
WHERE OrdMod <> OrdTrad;
```

## Concatenación
### + (signo más)

```python
%%sql
SELECT IDEmpleado, Nombre + ' ' + Apellido AS nombrecompleto
FROM Empleados;
```

### || (doble pipe)

```python
%%sql
SELECT TOP(10) IDCliente, Pais, Region, Ciudad,
Pais || ',' || Region || ',' || Ciudad AS ubicacion
FROM Clientes;
```
