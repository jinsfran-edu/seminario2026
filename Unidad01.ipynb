## CASE

```python
%%sql 
SELECT TOP(10) IDProducto, NombreProducto, IDCategoria,
CASE IDCategoria
WHEN 1 THEN 'Bebidas'
WHEN 2 THEN 'Condimentos'
WHEN 3 THEN 'Confites'
WHEN 4 THEN 'Lacteos'
WHEN 5 THEN 'Granos/Cereales'
WHEN 6 THEN 'Avicola'
WHEN 7 THEN 'Frutas'
WHEN 8 THEN 'Mariscos'
ELSE 'Categoria Desconocida'
END AS NombreCategoria
FROM Productos
```

```python
%%sql
SELECT TOP(10) IDPedido, IDCliente, val,
CASE
WHEN val < 1000.00 THEN 'Menos de 1000'
WHEN val BETWEEN 1000.00 AND 3000.00 THEN 'Entre 1000 y 3000'
WHEN val > 3000.00 THEN 'Mas de 3000'
ELSE 'Desconocido'
END AS CategoriaValor
FROM ValoresPedido
```

```python
%%sql
SELECT IDCliente, NombreEmpresa
FROM Clientes
ORDER BY IDCliente COLLATE Traditional_Spanish_100_CI_AI_SC_UTF8
OFFSET 13 ROWS FETCH NEXT 1 ROWS ONLY;
```

```python
%%sql
SELECT IDCliente, NombreEmpresa, OrdMod, OrdTrad
FROM (SELECT IDCliente, NombreEmpresa,
ROW_NUMBER() OVER 
(ORDER BY IDCliente COLLATE Traditional_Spanish_100_CI_AI_SC_UTF8) AS OrdTrad,
ROW_NUMBER() OVER 
(ORDER BY IDCliente COLLATE Modern_Spanish_100_CI_AI_SC_UTF8) AS OrdMod
FROM Clientes) AS D
WHERE OrdMod <> OrdTrad;
```

### QUOTENAME

```python
%%sql
SELECT QUOTENAME('abc[]def');
```

```python
%%sql
SELECT 'sin delimitador' as delimitador, QUOTENAME('Mi valor')       as salida, 1 as orden
UNION
SELECT ''' delimitador' as delimitador, QUOTENAME('Mi valor', '''') as salida, 2 as orden
UNION
SELECT '" delimitador'  as delimitador, QUOTENAME('Mi valor', '"')  as salida, 3 as orden
UNION
SELECT '[] delimitador' as delimitador, QUOTENAME('Mi valor', '[]') as salida, 4 as orden
UNION
SELECT '() delimitador' as delimitador, QUOTENAME('Mi valor', '()') as salida, 5 as orden
UNION
SELECT '<> delimitador' as delimitador, QUOTENAME('Mi valor', '<>') as salida, 6 as orden
UNION
SELECT '{} delimitador' as delimitador, QUOTENAME('Mi valor', '{}') as salida, 7 as orden
UNION
SELECT '` delimitador'  as delimitador, QUOTENAME('Mi valor', '`')  as salida, 8 as orden
ORDER BY orden
```
