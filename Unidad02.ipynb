```python
SELECT @@OPTIONS AS [Valor de @@OPTIONS];
SELECT
    GET_BIT(@@OPTIONS, 0) /* 1     */ AS [DISABLE_DEF_CNST_CHK] 
    , GET_BIT(@@OPTIONS, 1) /* 2     */ AS [IMPLICIT_TRANSACTIONS] 
    , GET_BIT(@@OPTIONS, 2) /* 4     */ AS [CURSOR_CLOSE_ON_COMMIT] 
    , GET_BIT(@@OPTIONS, 3) /* 8     */ AS [ANSI_WARNINGS] 
    , GET_BIT(@@OPTIONS, 4) /* 16    */ AS [ANSI_PADDING] 
    , GET_BIT(@@OPTIONS, 5) /* 32    */ AS [ANSI_NULLS] 
    , GET_BIT(@@OPTIONS, 6) /* 64    */ AS [ARITHABORT] 
    , GET_BIT(@@OPTIONS, 7) /* 128   */ AS [ARITHIGNORE] 
    , GET_BIT(@@OPTIONS, 8) /* 256   */ AS [QUOTED_IDENTIFIER] 
    , GET_BIT(@@OPTIONS, 9) /* 512   */ AS [NOCOUNT] 
    , GET_BIT(@@OPTIONS, 10) /* 1024  */ AS [ANSI_NULL_DFLT_ON] 
    , GET_BIT(@@OPTIONS, 11) /* 2048  */ AS [ANSI_NULL_DFLT_OFF] 
    , GET_BIT(@@OPTIONS, 12) /* 4096  */ AS [CONCAT_NULL_YIELDS_NULL] 
    , GET_BIT(@@OPTIONS, 13) /* 8192  */ AS [NUMERIC_ROUNDABORT] 
    , GET_BIT(@@OPTIONS, 14) /* 16384 */ AS [XACT_ABORT] 
GO
```

```python
IF YEAR(SYSDATETIME()) <> YEAR(DATEADD(day, 1, SYSDATETIME()))
PRINT 'Hoy es el último día del año.'; 
ELSE
PRINT 'Hoy no es el último día del año.';
```

```python
IF YEAR(SYSDATETIME()) <> YEAR(DATEADD(day, 1, SYSDATETIME()))
    PRINT 'Hoy es el último día del año.'; 
ELSE
    IF MONTH(SYSDATETIME()) <> MONTH(DATEADD(day, 1, SYSDATETIME()))
        PRINT 'Hoy es el último día del mes, pero no es el último día del año.'; 
    ELSE
        PRINT 'Hoy no es el último día del mes.';
```

```python
IF EXISTS (
SELECT s.name AS SchemaName, t.name AS TableName
FROM sys.schemas s JOIN sys.tables t ON s.schema_id = t.schema_id
WHERE s.name = 'dbo' AND t.name = 'Prueba1'
)
DROP TABLE Prueba1;
```

```python
IF NOT EXISTS (
SELECT s.name AS SchemaName, t.name AS TableName
FROM sys.schemas s JOIN sys.tables t ON s.schema_id = t.schema_id
WHERE s.name = 'dbo' AND t.name = 'Prueba1'
)
BEGIN
    PRINT 'Tabla dbo.Prueba1 no encontrada';
    PRINT 'Creando Tabla';
    CREATE TABLE Prueba1(Col1 int PRIMARY KEY);
END
ELSE
PRINT 'WARNING: Saltando CREATE. Tabla existe';
```
